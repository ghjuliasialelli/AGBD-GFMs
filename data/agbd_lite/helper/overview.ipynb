### Overview of how to load the AGBD-Lite data

Note that the already randomized samples in the dataset are stored contiguously in chunks of size 32. Thus, it is highly efficient to load every chunk at a time.

```python
from dataset import GEDIDataset
from omegaconf import OmegaConf
from torch.utils.data import DataLoader
```

```python
# Read the config file
conf = OmegaConf.load("config.yaml")

# Create the dataset
ds = GEDIDataset(mode = 'val', args = conf)

# Create the dataloader
data_loader = DataLoader(dataset = ds,
                        batch_size = 8, # each batch loads conf.lite_chunk_size elements already
                        shuffle = True,
                        num_workers = 8)

# Iterate through the dataloader
for batch_samples in data_loader:
    
    # Parse the batch
    features, agbs, biomes, regions = batch_samples
    print(features.shape, agbs.shape, biomes.shape, regions.shape)
    
    # Put the batches together
    features = features.flatten(0, 1)
    agbs = agbs.flatten(0, 1)
    biomes = biomes.flatten(0, 1)
    regions = regions.flatten(0, 1)
    print(features.shape, agbs.shape, biomes.shape, regions.shape)
    
    break
```
